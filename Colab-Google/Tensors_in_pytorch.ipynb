## Core Features of PyTorch

PyTorch is an open-source machine learning framework that is widely used for deep learning applications. Here are some of its core features:

1.  **Dynamic Computation Graph (Define-by-Run):** Unlike frameworks like TensorFlow 1.x which use static graphs, PyTorch uses a dynamic computation graph. This means the graph is built on-the-fly as operations are executed, allowing for more flexibility in designing complex models, especially those with variable-length inputs or dynamic control flow.

2.  **Automatic Differentiation (Autograd):** PyTorch's `autograd` engine automatically computes gradients for all tensor operations. This is crucial for training neural networks using backpropagation, as it eliminates the need for manual gradient calculation.

3.  **Pythonic Interface:** PyTorch is designed to be highly intuitive and integrates well with the Python ecosystem. Its API is similar to NumPy, making it easier for Python developers to learn and use.

4.  **GPU Acceleration:** PyTorch provides seamless support for GPU acceleration using CUDA, allowing for significantly faster computations, which is essential for training large deep learning models.

5.  **Extensive Library and Ecosystem:** PyTorch offers a rich set of modules, functions, and tools for various machine learning tasks, including `torch.nn` for neural network layers, `torch.optim` for optimizers, and `torchvision`, `torchaudio`, `torchtext` for domain-specific data handling and models.

6.  **Eager Mode Execution:** Operations in PyTorch are executed immediately, which simplifies debugging and makes the development process more interactive.

7.  **TorchScript:** For deployment and performance optimization, PyTorch offers TorchScript, a way to create serializable and optimizable models from standard PyTorch code. It allows for a transition from eager mode to a graph-based representation.

8.  **Distributed Training:** PyTorch includes robust features for distributed training, enabling models to be trained across multiple GPUs and machines, which is vital for handling massive datasets and model sizes.

## PyTorch Tensors vs. NumPy Arrays

Tensors are the fundamental data structure in PyTorch, serving a similar purpose to NumPy arrays but with crucial distinctions tailored for deep learning:

*   **Deep Learning Optimization:** PyTorch tensors are specifically designed and optimized for the operations commonly encountered in neural networks. This includes efficient handling of large multi-dimensional data and seamless integration with deep learning algorithms.
*   **GPU Acceleration:** One of the most significant advantages of PyTorch tensors is their ability to leverage GPU (Graphics Processing Unit) acceleration. Tensors can be easily moved to and operated on GPUs, enabling massive parallel computation. This is vital for significantly speeding up the training and inference of deep learning models, a capability that NumPy arrays lack as they are primarily CPU-bound.
*   **Automatic Differentiation (Autograd):** PyTorch tensors come with an built-in automatic differentiation engine, `autograd`. This engine can automatically track operations performed on tensors and compute their gradients (derivatives). This feature is fundamental for backpropagation, the algorithm used to train neural networks, as it eliminates the need for manual gradient calculation.
*   **Data Types:** While both support various numerical data types, PyTorch tensors often provide more specific and optimized data types relevant to deep learning (e.g., `torch.float32`, `torch.int64`, `torch.bfloat16`).
*   **Pythonic Integration:** PyTorch's API for tensors is very similar to NumPy's, making it familiar and easy for Python developers to adopt. This allows for a smooth transition for those already comfortable with NumPy.

In essence, you can conceptualize a PyTorch tensor as a NumPy array that has been enhanced with critical functionalities for deep learning, particularly GPU support and automatic differentiation.

## Tensor Operations on GPU

```python
if torch.cuda.is_available():
    device = torch.device("cuda")
    print(f"Using GPU: {torch.cuda.get_device_name(device)}")
else:
    device = torch.device("cpu")
    print("No GPU found, using CPU.")

# Create a tensor on the chosen device
tensor_on_device = torch.tensor([1.0, 2.0, 3.0]).to(device)
print(f"\nTensor on {device}:\n{tensor_on_device}")

# Perform a simple operation on the tensor
result = tensor_on_device + 5
print(f"\nResult of tensor + 5 on {device}:\n{result}")
```

```python
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
device
```

```python
# Creating a new tensor on GPU
torch.rand((2,3), device = device)
```

```python
# moving an existing tensor to GPU
a = torch.rand((2,3))
a
```

```python
a.to(device=device)
```

```python
import time

# Define the size of the matrices
size = 10000 #  Large size for performance comparison

# Create random matrices on CPU
matrix_cpu_a = torch.rand(size, size)
matrix_cpu_b = torch.rand(size, size)



# Measure time for CPU operation
start_time_cpu = time.time()
result_cpu = torch.matmul(matrix_cpu_a, matrix_cpu_b)
end_time_cpu = time.time() - start_time_cpu


print(f"Time on CPU: {end_time_cpu:.4f} seconds")

# Create random matrices on GPU
matrix_gpu_a = matrix_cpu_a.to(device)
matrix_gpu_b = matrix_cpu_b.to(device)

# Measure time on GPU
start_time_gpu = time.time()
result_gpu = torch.matmul(matrix_gpu_a, matrix_gpu_b)
torch.cuda.synchronize()       # Ensure all GPU operations are complete
end_time_gpu = time.time() - start_time_gpu


print(f"Time on GPU: {end_time_gpu:.4f} seconds")

# Compare Results
print("\nSpeedup (CPU time / GPU time):", end_time_cpu / end_time_gpu)
```
